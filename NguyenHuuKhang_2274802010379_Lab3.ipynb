```python
import pandas as pd
import numpy as np
import unicodedata
import re
from google.colab import files

print("=" * 80)
print("UPLOAD FILE CSV VÀO GOOGLE COLAB")
print("=" * 80)

uploaded = files.upload()
csv_file = list(uploaded.keys())[0]

print(f"\n✓ File đã upload: {csv_file}")


column_names = ["Id", "Name", "Age", "Weight", "m0006", "m0612", "m1218", "f0006", "f0612", "f1218"]

df = pd.read_csv(
    csv_file,
    names=column_names,
    engine='python',
    on_bad_lines='skip',
    encoding='utf-8-sig'
)

print("\n" + "=" * 80)
print("PROBLEM 1: Load dữ liệu từ CSV (missing header)")
print("=" * 80)
print(f"Shape: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
```

```python
print("\n" + "=" * 80)
print("PROBLEM 2: Split Name → Firstname + Lastname")
print("=" * 80)

df[['Firstname', 'Lastname']] = df['Name'].str.split(expand=True, n=1)

df = df.drop('Name', axis=1)

print("\nDataFrame sau split Name:")
print(df[['Id', 'Firstname', 'Lastname', 'Age', 'Weight']].head())
```

```python
print("\n" + "=" * 80)
print("PROBLEM 3: Chuẩn hóa Weight - chuyển lbs → kg")
print("=" * 80)

weight = df['Weight']

for i in range(len(weight)):
    x = str(weight[i])

    if "lbs" in x[-3:]:
        x = x[:-3]
        float_x = float(x)
        y = int(float_x / 2.2)
        y = str(y) + "kg"
        weight[i] = y

    elif "kg" in x[-2:]:
        weight[i] = x

    elif x == 'nan' or x == '-':
        weight[i] = np.nan

df['Weight'] = weight

print("\nWeight values sau chuẩn hóa (chuyển lbs → kg):")
print(df[['Id', 'Firstname', 'Lastname', 'Weight']].head(10))
```

```python
print("\n" + "=" * 80)
print("PROBLEM 4: Xóa rows rỗng (NaN)")
print("=" * 80)

print(f"\nBefore dropna: {len(df)} rows")

df.dropna(how="all", inplace=True)

print(f"After dropna: {len(df)} rows")
print("\nDataFrame sau xóa rows rỗng:")
print(df)
```

```python
print("\n" + "=" * 80)
print("PROBLEM 5: Drop duplicates (dựa trên subset columns)")
print("=" * 80)

print(f"\nBefore drop duplicates: {len(df)} rows")

# Drop duplicates dựa trên subset: Firstname, Lastname, Age, Weight
df = df.drop_duplicates(subset=['Firstname', 'Lastname', 'Age', 'Weight'])

print(f"After drop duplicates: {len(df)} rows")
print("\nDataFrame sau xóa duplicates:")
print(df)
```

```python
print("\n" + "=" * 80)
print("PROBLEM 6: Xóa non-ASCII characters")
print("=" * 80)

df['Firstname'] = df['Firstname'].replace({r'[^\x00-\x7F]+': ''}, regex=True)
df['Lastname'] = df['Lastname'].replace({r'[^\x00-\x7F]+': ''}, regex=True)

print("\nFirstname & Lastname sau xóa non-ASCII:")
print(df[['Firstname', 'Lastname']])
```

```python
print("\n" + "=" * 80)
print("PROBLEM 7: Missing Values - Các chiến lược xử lý")
print("=" * 80)

print("\n📊 Missing Values Statistics:")
print(df[['Age', 'Weight']].isnull().sum())

print("\nCác chiến lược xử lý missing values:")
print("  a. Deletion: Xóa records có missing values")
print("  b. Dummy substitution: Thay bằng dummy value (e.g., 0)")
print("  c. Mean substitution: Thay bằng mean")
print("  d. Frequent substitution: Thay bằng giá trị phổ biến nhất")
print("  e. Improve the data collector: Liên hệ clients")

print("\n✓ Chọn chiến lược: Mean substitution + Deletion (nếu thiếu cả 2)")

df['Weight_numeric'] = df['Weight'].str.extract(r'(\d+)').astype(float)

age_mean = df['Age'].mean()
weight_mean = df['Weight_numeric'].mean()

print(f"\nAge mean: {age_mean:.2f}")
print(f"Weight mean: {weight_mean:.2f}")


print("\n⚙️  Xử lý missing values:")
print(f"  - Rows thiếu Age: {df['Age'].isnull().sum()}")
print(f"  - Rows thiếu Weight: {df['Weight'].isnull().sum()}")

missing_both = df[(df['Age'].isnull()) & (df['Weight'].isnull())]
print(f"  - Rows thiếu cả Age và Weight: {len(missing_both)}")

df = df.drop(missing_both.index)

df['Age'] = df['Age'].fillna(age_mean)

df['Weight_numeric'] = df['Weight_numeric'].fillna(weight_mean)
df['Weight'] = df['Weight_numeric'].astype(int).astype(str) + "kg"

df = df.drop('Weight_numeric', axis=1)

print("\nDataFrame sau xử lý missing values:")
print(df[['Id', 'Firstname', 'Lastname', 'Age', 'Weight']])
```

```python
print("\n" + "=" * 80)
print("PROBLEM 8: Unpivot dữ liệu - tách heart rate columns")
print("=" * 80)

print("\nBefore unpivot shape:", df.shape)

id_vars = ['Id', 'Age', 'Weight', 'Firstname', 'Lastname']
value_vars = ['m0006', 'm0612', 'm1218', 'f0006', 'f0612', 'f1218']

df_melted = pd.melt(
    df[id_vars + value_vars],
    id_vars=id_vars,
    value_vars=value_vars,
    var_name='sex_and_time',
    value_name='PulseRate'
)

print("\nAfter melt shape:", df_melted.shape)

def extract_sex_and_time(sex_and_time_str):
    """Extract Sex và Time từ sex_and_time"""

    sex = sex_and_time_str[0]
    sex_full = 'male' if sex == 'm' else 'female'

    time_code = sex_and_time_str[1:]

    hours_lower = int(time_code[:2])
    hours_upper = int(time_code[2:])

    time_range = f"{hours_lower:02d}-{hours_upper:02d}"

    return sex, sex_full, time_range

sex_data = df_melted['sex_and_time'].apply(extract_sex_and_time)

df_melted['Sex_code'] = sex_data.apply(lambda x: x[0])
df_melted['Sex'] = sex_data.apply(lambda x: x[1])
df_melted['Time'] = sex_data.apply(lambda x: x[2])

df_melted = df_melted.drop('sex_and_time', axis=1)

print("\nColumns sau extract:")
print(df_melted.columns.tolist())

print("\nFirst 10 rows sau unpivot:")
print(df_melted.head(10))
```

```python
output_filename = "patient_heart_rate_cleaned.csv"
df_melted.to_csv(output_filename, index=False)

print(f"\n✓ Cleaned data saved: {output_filename}")
print(f"✓ Total rows: {len(df_melted)}")
print("✓ Download từ file manager hoặc Google Drive")

# Optional: Save to Google Drive
try:
    from google.colab import drive
    drive.mount('/content/drive')
    df_melted.to_csv('/content/drive/My Drive/patient_heart_rate_cleaned.csv', index=False)
    print("✓ Cũng lưu vào Google Drive: My Drive/patient_heart_rate_cleaned.csv")
except:
    pass
```
